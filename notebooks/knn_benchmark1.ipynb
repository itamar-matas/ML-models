# Test 1 - Iris Species

#### Step 1 - imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import sys
import os

sys.path.append(os.path.abspath(os.path.join('..')))
```

```python
from models import KNNClassifier
import metrics as mtr
import preprocessing as prp
import visuals as vis
```

#### Step 2 - data handling

```python
df = pd.read_csv('../data/raw/Iris.csv')
df.head()
```

```python
df['Species'] = df['Species'].astype('category').cat.codes
df.head()
```

```python
Y = df['Species'].values
X = df.drop(['Species', 'Id'], axis=1).values

X_train, X_test, Y_train, Y_test = prp.train_test_split(X, Y)
```

```python
minmax = prp.normalize.minmax()

X_train = minmax.fit_transform(X_train)
X_test  = minmax.transform(X_test)
```

#### Step 3 - visualization

```python
mapping = np.array(['Setosa', 'Versicolor', 'Virginica'])
inverse_transform = lambda labels: mapping[labels]

vis.style(style='whitegrid')

vis.plot_feature_relationships(X_train, Y_train, feature_labels=['Sepal Length [cm]','Sepal Width [cm]','Petal Length [cm]','Petal Width [cm]'], inverse_transform=inverse_transform)
```

#### Step 4 - Training and evaluation

```python
knn1  = KNNClassifier(k=1)
knn3  = KNNClassifier(k=3)
knn5  = KNNClassifier(k=5)
knn10 = KNNClassifier(k=10)
knn20 = KNNClassifier(k=20)
knn50 = KNNClassifier(k=50)


knn1.fit(X_train, Y_train)
knn3.fit(X_train, Y_train)
knn5.fit(X_train, Y_train)
knn10.fit(X_train, Y_train)
knn20.fit(X_train, Y_train)
knn50.fit(X_train, Y_train)
```

```python
Y1_pred,  Y1_confs  = knn1.predict(X_test)
Y3_pred,  Y3_confs  = knn3.predict(X_test)
Y5_pred,  Y5_confs  = knn5.predict(X_test)
Y10_pred, Y10_confs = knn10.predict(X_test)
Y20_pred, Y20_confs = knn20.predict(X_test)
Y50_pred, Y50_confs = knn50.predict(X_test)

accuracies  = [round(mtr.accuracy_score(Y_test, y_pred), 4) for y_pred in [Y1_pred, Y3_pred, Y5_pred, Y10_pred, Y20_pred, Y50_pred]]
confidences = [round(y_confs.mean(), 4) for y_confs in [Y1_confs, Y3_confs, Y5_confs, Y10_confs, Y20_confs, Y50_confs]]

print("k=1:  " + str(accuracies[0]) + ", avg confidence: " + str(confidences[0]))
print("k=3:  " + str(accuracies[1]) + ", avg confidence: " + str(confidences[1]))
print("k=5:  " + str(accuracies[2]) + ", avg confidence: " + str(confidences[2]))
print("k=10: " + str(accuracies[3]) + ", avg confidence: " + str(confidences[3]))
print("k=20: " + str(accuracies[4]) + ", avg confidence: " + str(confidences[4]))
print("k=50: " + str(accuracies[5]) + ", avg confidence: " + str(confidences[5]))
```

```python
plt.plot([1, 3, 5, 10, 20, 50], accuracies, label="accuracy", c='purple', marker='o')
plt.plot([1, 3, 5, 10, 20, 50], confidences, label="confidence", c='teal',  marker='o')

plt.title("Accuracy an Confidence for K")

plt.legend()
plt.show()
```

```python
conf_mats = [mtr.confusion_matrix(Y_test, y_pred) for y_pred in [Y1_pred, Y3_pred, Y5_pred, Y10_pred, Y20_pred, Y50_pred]]
Ks = [1, 3, 5, 10, 20, 50]

vis.plot_confusion_matrices(
    conf_mats,
    ["Satosa", "Versicolor", "Virginica"],
    titling=lambda i: f"Confusion Matrix for K = {Ks[i]}"
)
```
